```python
# Load  Datasets

import pandas as pd

products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")
customers = pd.read_csv("customers.csv")

```

```python
products.head()

```

```python
orders.head()
```

```python
customers.head()
```

```python
combined = pd.merge(orders, products, on='product_id', how='left')
combined.head()
```

```python
cd = pd.merge(combined, customers, on='customer_id', how='left')
cd.head()
```

```python
# Export cd to CSV with specified filename

cd.to_csv('master_data.csv', index=False)
```

```python
# Convert order_date column to datetime format

cd['order_date'] = pd.to_datetime(cd['order_date'])
```

```python
cd
```

```python
cd.info()
```

```python
cd['revenue'] = cd['price'] * cd['quantity']
```

```python
cd
```

```python
cd['order_month'] = cd['order_date'].dt.month_name()
```

```python
cd
```

```python
# Standardize all column headings to lower case

cd.columns = cd.columns.str.lower()
```

```python
cd.head()
```

```python
total_revenue = cd['revenue'].sum()
total_revenue
```

```python
# Top 5 products by total revenue (descending)

top_products = (
    cd.groupby('product_id')['revenue']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_products
```

```python
# Top 5 products by total quantity sold (descending)

top_quantity = (
    cd.groupby('product_id')['quantity']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_quantity
```

```python
# Calculate AOV (Average Order Value)

total_revenue = cd['revenue'].sum()
total_orders = cd['order_id'].nunique()

aov = round(total_revenue / total_orders, 2)
aov
```

```python
# Count total number of unique customers

unique_customers = cd['customer_id'].nunique()
unique_customers
```

```python
# Total number of orders per region

orders_by_region = (
    cd.groupby('region')['order_id']
    .nunique()
)

orders_by_region
```

```python
# Total revenue per region

revenue_by_region = (
    cd.groupby('region')['revenue']
    .sum()
)

revenue_by_region
```

```python
# Top 5 customers by total spend (revenue)

top5_customers = (
    cd.groupby('customer_id')['revenue']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top5_customers
```

```python
# Group by category and compute totals

category_summary = (
    cd.groupby('category')
    .agg(
        total_revenue=('revenue', 'sum'),
        total_quantity=('quantity', 'sum')
    )
)

category_summary


# Category with highest total revenue
top_category_by_revenue = category_summary['total_revenue'].idxmax()
top_category_by_revenue
```

```python
# Total revenue per month

monthly_revenue = (
    cd.groupby('order_month')['revenue']
    .sum()
)

monthly_revenue


# Highest monthly revenue (rounded to 2 decimals)
high_monthly_revenue = round(monthly_revenue.max(), 2)
high_monthly_revenue
```
